```python
# Find the minimum, maximum, and average of total production.
print("min",data['Total Production'].min())
print("max",data['Total Production'].max())
print("average",data['Total Production'].mean())
```

```python
# Which country has the highest total production in a single record?
data.loc[data['Total Production'].idxmax(), 'Country']
```

```python
# Are there any rows where total production is missing or zero?
missing_or_zero=data[(data['Total Production'].isnull()) | (data['Total Production'] == 0)]
missing_or_zero
```

```python
# What is the average import price across all records?
data['Import Price'].mean()
```

```python
# What is the total coffee production for each year?
data.groupby('Year')['Total Production'].sum()
```

```python
# Which are the top 5 countries based on total production?
data.groupby('Country')['Total Production'].sum().sort_values(ascending=False).head(5)
```

```python
# Which country has the highest average export price?\
data.groupby('Country')['Export Price'].mean().sort_values(ascending=False).head(1)
```

```python
# Does domestic consumption exceed total production for any country?
data[data['Domestic Consumption'] > data['Total Production']]
```

```python
# Which country shows the highest growth in total production over the years?
data.groupby('Country')['Total Production'].sum().sort_values(ascending=False).head(1)
```

```python
# Is there any relationship between export price and total production
plt.scatter(data['Export Price'], data['Total Production'])
```
